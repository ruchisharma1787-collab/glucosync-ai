```python
import pandas as pd
df = pd.read_csv("../data/diabetes_cleaned.csv")
```

```python
df["Glucose"].mean()
```

```python
df.groupby("Outcome")["Glucose"].mean()
```

```python
df.groupby("Outcome")["BMI"].mean()
```

```python
df.groupby("Outcome")["Glucose"].agg(["mean","min","max","count"])
```

```python
df.groupby("Outcome")[["Glucose" , "BMI" , "Age"]].mean()
```

```python
df["Age_Group"] = pd.cut(df["Age"], bins=[20, 30, 40, 50, 100])
df.pivot_table(values="Glucose", index="Age_Group", columns="Outcome", aggfunc="mean")
```

```python
df.pivot_table(index = "Outcome", values = ["Glucose", "BMI"], aggfunc = ["mean", "min", "max", "count"])
```
