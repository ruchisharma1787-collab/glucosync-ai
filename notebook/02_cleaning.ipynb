```python
import pandas as pd 
import numpy as np 
# 1. Data load 
df = pd.read_csv("../data/diabetes.csv")
# 2. Finding missing vales 
df.isna().sum()
# 3. Hidden Zeros Column 
cols = ["Glucose" , "BloodPressure" , "SkinThickness","Insulin", "BMI"]
# 4. 0 to NAN (here nan tells pandas that this value is missing)
df[cols] = df[cols].replace(0,np.nan)
# 5. finding missing values again 
print("Missing values before filling:")
print(df.isna().sum())
# 6. Missing values of every column is filled with median 
for c in cols:
    df[c] =df[c].fillna(df[c].median())
# 7. Confirm  once again if all values are filled
print("\n Missing values after filling:")
print(df.isna().sum())
# 8. Duplicate rows check and remove
print("\nDuplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
# 9. Save the cleaned data into another file 
df.to_csv("../data/diabetes_cleaned.csv" , index = False)
print("\nCleaned file saved as diabetes_cleaned.csv")


```
